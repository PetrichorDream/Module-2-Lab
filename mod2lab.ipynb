```python
# Author: Trevor S
# File Name: deans_honor_roll.py
# Desc: Accepts student names and GPAs to check if they qualify
#       for the Dean's list (GPA >= 3.5) or Honor Roll (GPA >= 3.25).
#       Stops running when 'ZZZ' is entered as the last name.

while True:
  last_name = input("Enter student's last name (or 'ZZZ' to quit): ")

  if last_name.upper() == 'ZZZ':
    print("Exiting")
    break

  first_name = input("Enter student's first name: ")

  try:
    gpa = float(input("Enter student's GPA: "))
  except ValueError:
    print("Invalid input. Please enter a float value for GPA like (3.34) to continue.\n")
    continue

  # Check standing
  if gpa >= 3.5:
    print(f"{first_name} {last_name} has made the Dean's List.\n")
  elif gpa >= 3.25:
    print(f"{first_name} {last_name} has made Honor Roll.\n")
  else:
    print(f"{first_name} {last_name} did not qualify for the Dean's List or Honor Roll.\n")
```
